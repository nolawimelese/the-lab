fancy indexing

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

x = rng.integers(0,10,(6,5))
print(x)
```

```python
# manual indexing
print(x[0][3], x[2][3], x[1][4])

# fancy indexing
x_ind = [0, 2, 1]
y_ind = [3, 3, 4]
print(x[x_ind, y_ind])

# combined indexing
print(x[2, [0, 3, 4]])

# fancy indexing and slicing
print(x[2:, [3,2,1]])
```

```python
# binning

x = np.linspace(0,9,10)
print(x)
data = rng.integers(0,10,50)
print(data)
counts = np.zeros_like(x)
i = np.searchsorted(x,data)
np.add.at(counts,i,1)
plt.plot(x,counts,drawstyle='steps')
```
