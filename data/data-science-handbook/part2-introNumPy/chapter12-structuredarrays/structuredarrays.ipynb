Structured arrays can be used to have fixed arrays of compound data types

```python
import numpy as np
```

```python
data = np.zeros(4, dtype={'names':('name','age','hobby'), 
                          'formats':('U10','i4','U10')})
data.dtype
data
```

```python
data[0] = ('Nolawi', '19', 'Mario Kart')
data[1] = ('Jim', '30', 'Driving')
data[2] = ('Shelby', '24', 'Dancing')

print(data)

print(data[0]['name'])
print(data[1]['age'])
print(data[2]['hobby'])
```

```python
# bit of masking

print(data[data['hobby'] == 'Mario Kart'])
```

```python
print(np.dtype('b'))   # 8-bit integer

print(np.dtype('i8'))  # signed integer

print(np.dtype('u8'))  # unsigned integer

print(np.dtype('f8'))  # floating point

print(np.dtype('c16')) # complex floating point

print(np.dtype('S5'))  # string
print(np.dtype('a5'))  # string

print(np.dtype('U5'))  # unicode string

print(np.dtype('V10')) # raw data (void)
```
