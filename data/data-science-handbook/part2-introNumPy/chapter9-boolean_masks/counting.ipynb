```python
import numpy as np
import seaborn as sns

print(sns.get_dataset_names())
```

```python
df = sns.load_dataset('car_crashes').head(10)
print(df)
len(df)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')

plt.figure(figsize=(8, 4.5))
sns.barplot(data=df, x='abbrev', y='total', color='#4269d0', width=0.65)

plt.xlabel('State')
plt.ylabel('Total crashes per billion miles')
plt.title('Total car crashes, first 10 states')
sns.despine(left=True)

plt.show()
```
