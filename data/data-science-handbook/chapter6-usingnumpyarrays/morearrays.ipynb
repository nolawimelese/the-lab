Notes on: Using NumPy arrays vs Python Lists

```python
import numpy as np
```

```python
def multiply_2(arr):
    output = np.empty(len(arr))
    for i in range(len(arr)):
        output[i] = arr[i] * 2
    return output
```

```python
rng = np.random.default_rng()

%timeit multiply_2(rng.integers(1,10, size = 500))

%timeit rng.integers(1,10,size=500) * 2
```
